# 05 Real-data model metrics

Metrics are MAE, RMSE, and R² on later harvest years. R² is the coefficient of determination, not classification accuracy. The 0.80 R² target is a research threshold, not a claim that has already been met. If farm-level Macheke yields are absent, the fit below uses FAOSTAT national yield and must not be described as a farm-level model.

```python
import sys
from pathlib import Path
ROOT = Path.cwd()
if not (ROOT / "src").is_dir():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))
import pandas as pd
DATA = ROOT / "data"

from models.real_training import WARNING, train_from_real_dataset
print(WARNING)
model_frame = pd.read_csv(DATA / "processed" / "modelling_dataset.csv")
result = train_from_real_dataset(model_frame, ROOT / "reports" / "model_reports")
print("status", result["status"])
print("chosen", result.get("chosen_model"))
print("test MAE", result.get("metrics", {}).get("mae"))
print("test RMSE", result.get("metrics", {}).get("rmse"))
print("test R2", result.get("metrics", {}).get("r2"))
pd.read_csv(ROOT / "reports" / "model_reports" / "model_comparison.csv")
```
