# 04 Real data EDA

Rainfall and temperature describe the Macheke study-area grid. Yield, when plotted, is FAOSTAT national Zimbabwe maize yield. A correlation here is not a Macheke farm finding.

```python
import sys
from pathlib import Path
ROOT = Path.cwd()
if not (ROOT / "src").is_dir():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))
import pandas as pd
DATA = ROOT / "data"

frame = pd.read_csv(DATA / "processed" / "capstone_dataset_real.csv")
columns = ["season_rainfall_mm", "avg_temperature_c", "max_temperature_c", "min_temperature_c", "heat_stress_days", "yield_t_ha", "soil_ph"]
print(frame[columns].describe())
print(frame[columns].corr(numeric_only=True))
```
